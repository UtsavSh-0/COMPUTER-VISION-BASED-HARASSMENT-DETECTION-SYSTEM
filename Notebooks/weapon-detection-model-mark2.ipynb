```python
import os
import yaml
import shutil
from ultralytics import YOLO
```

```python
dataset_root = r"../Data/weapon detection cctv v3 dataset.v1-weapon_detection_in_cctv.yolov8"

yaml_path = os.path.join(dataset_root, "data.yaml")

with open(yaml_path, 'r') as f:
    data = yaml.safe_load(f)

data['path'] = dataset_root
data['train'] = os.path.join(dataset_root, "train", "images")
data['val'] = os.path.join(dataset_root, "valid", "images")
data['test'] = os.path.join(dataset_root, "test", "images")

with open(yaml_path, 'w') as f:
    yaml.dump(data, f, sort_keys=False)

print("data.yaml updated successfully for YOLO training!\n")
for split in ['train', 'valid', 'test']:
    split_dir = os.path.join(dataset_root, split, 'images')
    print(f"{split.capitalize()} images: {len(os.listdir(split_dir)) if os.path.exists(split_dir) else 0}")
```

```python
model = YOLO("yolov8n.pt")
model.train(
    data=yaml_path,
    epochs=50,
    imgsz=640,
    batch=16,
    project='runs/detect',
    name='cctv_yolov8_training',
    patience=5,
    amp=False
)
```

```python
metrics = model.val()
print(metrics)
```

```python
results = model.predict(source=data['test'], save=True, conf=0.25)
```

```python
best_model_path = os.path.join('runs', 'detect', 'cctv_yolov8_training', 'weights', 'best.pt')
destination_path = 'cctv-weapon-model-mark2.pt'

if os.path.exists(best_model_path):
    shutil.copy(best_model_path, destination_path)
    print(f'Model successfully saved to {destination_path}')
else:
    print('Error: Could not find the trained model file.')
```
